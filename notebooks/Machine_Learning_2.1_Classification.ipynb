## The Heart disease dataset

For discovering classification, we're gonna use a rather classic dataset: the heart disease dataset. This dataset contains 13 symptoms and other attributes of patients that have been checked for a heart disease, such as their age, their cholesterol blood level, the type of pain they report, .... observed on 303 patients. These 13 attributes are called the *features*. And for each of these patients, we know if they have a heart disease or not: the field called *target* in the dataset. This field is generally called the *classes* of the problem, in this case we have two classes : 0 for healthy patients, and 1 for patients with a heart disease, hence it is called a *binary classification problem*.

As with linear regression, we generally denote the feature matrix by $X$, and the classes with $y$, where, in this dataset:


$$X \in \mathbb{R}^{303 \times 13}, y \in  \{0,1\}^{303}.$$
The goal of classification is to learn a function, or *classifier*, $f$ that approximates the true classes $y$ from the features $X$:


$$y \approx f(X).$$

However as $y$ is categorical in classification problems since it represents different classes, as opposed to a continuous value as in linear regression, we cannot use linear regression for solving these problems, and we need different models. 

Also note that classification is not always binary, and can classify features between more than 2 classes (will not be covered today, but the models that we will use here can be directly used with any number of classes).

To get a better idea of these concepts, read chapter 4.1 and 4.2 from the *Introduction to Statistical Learning* book.

## Preprocessing data

We have seen that we have categorical features, for example the *cp* feature that describes the chest pain type reported with number from 0 to 3:
0=typical angina; 1=atypical angina; 2=non-anginal pain; 3=asymptomatic.

However these are *qualitative* categories, and there is no notion of distance between these numbers: an atypical angina is not 2 times a typical angina! 

To handle that, we are going to create what we call *dummy variables* (=*one hot encoding*) that is, for each value 0, 1, 2, and 3, we are going to create a new feature which is a 0-1 feature only, and has a 1 value only for its own value. However we will do that only for *multicategorical* features : the ones that have more than two distinct values, since binary features like *sex*, *exang* and *fbs* are already 0-1 features.

```python
multicategorical_features = ['cp', 'restecg', 'slope', 'ca', 'thal']
continuous_features = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']
```

To create dummy variables, have a look at the `get_dummies` function from pandas (this does the same thing as the `OneHotEncoder` but with nice column names).

```python
#TOFILL 
## Aplico get_dummies() solamente a las variables categóricas, no a las variables continuas para no perder su informacion numérica.

dataset_dummies = pd.get_dummies(dataset, columns=multicategorical_features, dtype=int)

print(dataset_dummies[continuous_features].head(),'\n')

print(dataset_dummies.describe().head())
```

## Classification with k-nearest neighbors, and classification metrics

One easy introduction to classification is to start with the k-neareast neighbors method (KNN). At training, it simply memorizes all the training samples features $X$ and classes $y$. At test time given the features of one sample $x'$, it identifies the $k$ training samples $x_i, i \in 1,\dots,k$ that are the closest to $x'$ (in euclidian distance), and assign the class $y'$ that is the most frequent among the k-neareast neighbor classes $y_i, i \in 1,\dots,k$.

So each test sample is assigned a probablity, for example of having a heart disease:


$$P(y' = 1 ) = \frac{1}{k} \sum_{ i \in 1,\dots,k} \mathbb{1}(y_i = 1) $$
    
where the indicator function $\mathbb{1}(y_i = 1) = 1$ if $y_i = 1$, else $\mathbb{1}(y_i = 1) = 0$. So the probability of having a heart disease is the proportion of the k-nearest train samples that have a heart disease.

For an intuitive explanation of KNNs, watch the first 1:45min of this vid (the rest of it that is about Voronoi partitions is not necessary): https://www.youtube.com/watch?v=UqYde-LULfs

You can also read pages 38-40 from the *Introduction to Statistical Learning*.

Compute the predictions of a k-nearest neighbors classifier with 5-fold crossvalidation, with 15 neighbors.
You can remind yourself about cross-validation here:
https://scikit-learn.org/stable/modules/cross_validation.html

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_predict

"""
En mi caso, de los 303 registros reservo el 30 % para test:

    Entrenamiento: aproximadamente 212 registros.
    Test: aproximadamente 91 registros.
    Validación cruzada: 5 particiones sobre los 212 registros de entrenamiento.

Importante: utilizamos X_train_scaled, no X_train, porque KNN calcula distancias entre observaciones. El escalado es especialmente importante para que variables como el colesterol o la edad no dominen el cálculo de distancias por sus diferentes escalas.
"""

n_neighbors = 15 # Defino la cantidad de vecinos a considerar "TALLA"

## Creo el Clasificador KNN

#TOFILL: 
## KNeighborsClassifier(n_neighbors=15): crea un clasificador que utiliza los 15 vecinos más cercanos para determinar la clase.
knn_clf = KNeighborsClassifier(n_neighbors=n_neighbors) # Creo el objeto(Clasificador) instanciando el constructor de la clase "KNeighborsClassifier" y le paso como parametro la cantidad de vecinos

## instanciando el constructor de la clase "cross_val_predict" y le paso como parametro El objeto clasificador, X_train_scaled, y_train,
## Obtener predicciones mediante validación cruzada de 5 folds
# cross_val_predict(): divide los datos de entrenamiento en 5 particiones
# y_cv_pred: almacenara las predicciones obtenidas para todos los registros de entrenamiento.
y_cv_pred = cross_val_predict(knn_clf, X_train_scaled, y_train, cv=5) # (cv=5): realiza cinco iteraciones. En cada una, entrena con cuatro particiones y predice sobre la restante.

print("Predicciones:", y_cv_pred)
print("Número de predicciones:", len(y_cv_pred))

```

Use these predictions to compute the accuracy of your 15-nearest neighbors model :

```python
## calcular la precisión (accuracy) del clasificador KNN con 15 vecinos
from sklearn.metrics import accuracy_score


#TOFILL
## Calculate Accurancy
accuracy = accuracy_score(y_train, y_cv_pred)

print(f"Accuracy: ",accuracy,"\n")

print(f"Accuracy: {accuracy * 100:.2f}%") # La precisión(Accurancy) representa la proporción de predicciones correctas respecto al total de predicciones.
```

$\approx$ 81%, pretty good for someone who doesn't know about medicine ! However accuracy is a quite limited view of the prediction abilities of your classifier. Indeed it doesn't make the difference between patients that have been incorrectly classified with a heart disease (because they were healthy) with patients that have been incorrectly classified as healthy (because they had a disease).

Let's look at the *confusion matrix* of the prediction, it gives use the number of patients that have been correctly classified as having a disease : the *true positives* (TP) ; the number of patients that have been correctly classified as not having a disease : the *true negatives* (TN); the number of patients that have been incorrectly classified as having a disease : the *false positives* (FP) ; and the number of patients that have been incorrectly classified as nothaving a disease : the *false negatives* (FN). To remember more easily, remark that the true/false refers to the *true* class of the test samples, whereas the positive/negative refers to the *predicted* class by the classifier.

The confusion matrix gives these four numbers in the following format:

|  |  |
|--|--|
|TN|FP|
|FN|TP|

The accuracy is computed by :


$$accuracy= \frac{TP + TN}{TP + TN + FP + FN}$$

One can also look at the accuracy of positive predictions, called the *precision*:


$$precision= \frac{TP }{TP + FP}$$

Or at the ratio of positive samples correctly detected by the classifier, called the *recall*:


$$recall= \frac{TP }{TP + FN}$$
     

These two metrics are often grouped together as a single one called the *f1-measure*:

$$F1=  2 \times \frac{precision \times recall}{precision + recall}$$

You can read more about it in chapter 3 of the book *Hands on Machine Learning with scikit-learn and tensorflow*.

So these metrics gives use different informations about our classifier predictive performances, we are gonna see in more details how below, first compute them on the results of your predictions :

```python
from sklearn.metrics import confusion_matrix,  precision_score, recall_score, accuracy_score, f1_score
#TOFILL

""" 
*** La matriz de confusión (Confusion Matrix): La matriz de confusión nos permite visualizar los aciertos y errores de nuestro modelo.
1. True Positive (TP) — Verdaderos positivos El modelo predice que el paciente tiene una enfermedad y efectivamente la tiene.
2. True Negative (TN) — Verdaderos negativos El modelo predice que el paciente está sano y efectivamente está sano.
3. False Positive (FP) — Falsos positivos El modelo predice que el paciente tiene una enfermedad, pero realmente está sano.
4. False Negative (FN) — Falsos negativos El modelo predice que el paciente está sano, pero realmente tiene una enfermedad.
Este error puede ser especialmente peligroso porque el paciente podría no recibir el tratamiento que necesita.
Truco para recordarlo:
    True = el modelo ha acertado.
    False = el modelo se ha equivocado.
    Positive = el modelo predice que existe la enfermedad.
    Negative = el modelo predice que no existe la enfermedad.
    
*** Accuracy (Exactitud): Indica qué porcentaje de predicciones ha realizado correctamente nuestro modelo.
*** Precision (Precisión): La precisión responde a esta pregunta:
    De todos los pacientes que nuestro modelo ha identificado como enfermos, ¿cuántos realmente están enfermos?
*** Recall (Sensibilidad): El recall responde a una pregunta diferente:
    De todos los pacientes que realmente están enfermos, ¿cuántos ha conseguido identificar correctamente nuestro modelo?
*** F1-score (F1-measure): Ahora tenemos dos métricas: 
    Precision: mide la fiabilidad de las predicciones positivas.
    Recall: mide la capacidad del modelo para detectar los casos positivos reales.
    El F1-score combina ambas métricas en un único valor. Cuanto más se acerque a 1, mejor será el equilibrio entre precision y recall.
    Es especialmente útil cuando queremos tener en cuenta ambas métricas simultáneamente.
"""

## Calculate the confusion matrix
conf_matrix = confusion_matrix(y_train, y_cv_pred)

print("Confusion Matrix:") 
print(conf_matrix)

## Calculate Accuracy
accuracy = accuracy_score(y_train, y_cv_pred) # Indica qué porcentaje de predicciones ha realizado correctamente nuestro modelo

## Calculate Precision
precision = precision_score(y_train, y_cv_pred) # De todos los pacientes que nuestro modelo ha identificado como enfermos, ¿cuántos realmente están enfermos?

# Calculate Recall
recall = recall_score(y_train, y_cv_pred) # De todos los pacientes que realmente están enfermos, ¿cuántos ha conseguido identificar correctamente nuestro modelo?

# Calculate F1-score
f1 = f1_score(y_train, y_cv_pred) # combina ambas métricas en un único valor Cuanto más se acerque a 1, mejor será el equilibrio entre precision y recall.

print("Accuracy:", accuracy) 
print("Precision:", precision) 
print("Recall:", recall) 
print("F1-score:", f1)

print("False Positives:", conf_matrix[0][1])

print("False Negatives:", conf_matrix[1][0])
```

## The precision/recall trade-off

We have assessed some measures that are applicables to the predicted classes, i.e. 0 or 1. Remember that KNN assigns a probability to each test sample, and simply predicts 1 if this probability is higher than > 0.5:


$$P(y' = 1 ) = \frac{1}{k} \sum_{ i \in 1,\dots,k} \mathbb{1}(y_i = 1)  > 0.5$$


Let us say we are here performing a first diagnosis test with our classifier, for further medical investigation if the prediction is positive. In this context, it is much more important to not say someone is healthy if he is not, rather than saying someone is sick if he is not (which can be discovered with later medical tests). In other words, we want to have a few false negatives, even if that implies having more false positives. This means we'd prefer to have a higher Recall, and thus a lower Precision. And that implies choosing a threshold that is lower than 0.5 for assigning the classes.


Use *method ='predict_proba'* to get the probability of each sample from the cross-validation, and print these probabilities:

```python
#TOFILL
#Print the probability associated with y_cv_proba:
# Get probabilities using cross-validation
y_cv_proba = cross_val_predict(knn_clf, X_train_scaled, y_train, cv=5, method='predict_proba')

""" 
¿Qué diferencia hay entre (predict y predict_proba)?
    ** predict() → devuelve la clase predicha (0 o 1).
    ** predict_proba() → devuelve la probabilidad de pertenecer a cada clase.
    En resumen: sustituimos la predicción de clases por las probabilidades 
    utilizando method='predict_proba' dentro de cross_val_predict().
"""

# Print probabilities
# print(y_cv_proba)

# Colores ANSI
GREEN = "\033[92m"
RED = "\033[91m"
RESET = "\033[0m"

# Recorrer las probabilidades de cada paciente
for proba in y_cv_proba:

    healthy = proba[0] * 100  # Probabilidad de estar sano
    disease = proba[1] * 100  # Probabilidad de estar enfermo

    print(
        f"Probabilité d'être en bonne santé : "
        f"{GREEN}{healthy:.2f}%{RESET} - "
        f"Probabilité d'être malade : "
        f"{RED}{disease:.2f}%{RESET}"
    )
```

`y_cv_proba` should be a vector of probabilities of being positive of all samples, now let's check that if we apply a 0.5 threshold, we obtain the same predictions as before:

```python
#By default KNN applies a 0.5 threshold to make its predictions:
""" 
    y_cv_proba[:, 1]: selecciona la probabilidad de tener una enfermedad cardíaca.
    > 0.5: comprueba si esa probabilidad supera el 50 %.
    .astype(int): convierte los valores booleanos en enteros (True = 1, False = 0).
    == y_cv_pred: compara las nuevas predicciones con las predicciones originales.
    .all(): comprueba si todas las predicciones coinciden.
    
    Si el RESULTADO ES "TRUE" significa que todas las predicciones obtenidas aplicando 
    el umbral de 0.5 coinciden con las predicciones originales de KNN.
    Una observación importante: el código del enunciado utiliza > 0.5, mientras que KNN 
    puede asignar una clase en situaciones de empate (por ejemplo, 7 vecinos sanos y 8 enfermos no es empate, 
    pero 7 y 8; un empate sería imposible con 15 vecinos). Con 15 vecinos no hay empate 
    entre dos clases, por lo que el umbral de 0.5 funciona correctamente.
"""
((y_cv_proba[:,1] > 0.5).astype(int) == y_cv_pred).all()
```

Let's plot the different values of Precision and Recall for different thresholds:

```python
#Precision recall tradeoff
from sklearn.metrics import precision_recall_curve

def plot_precision_recall_vs_threshold(precisions, recalls, thresholds):
    plt.plot(thresholds, precisions[:-1], "b--", label="Precision")
    plt.plot(thresholds, recalls[:-1], "g-", label="Recall")
    plt.xlabel("Threshold")
    plt.legend(loc="upper left")
    plt.ylim([0, 1])

#TOFILL
""" 
    Si la probabilidad de enfermedad es superior a 0.5 → paciente enfermo (1).
    Si es inferior o igual a 0.5 → paciente sano (0).
    
    ** Precision: proporción de predicciones positivas que son realmente positivas.
    ** Recall: proporción de pacientes enfermos que conseguimos detectar.
"""
## Calculate precision, recall and thresholds
precisions, recalls, thresholds = precision_recall_curve( y_train, y_cv_proba[:, 1] ) # Calculo las diferentes precisiones, recalls y thresholds

""" 
    precisions: valores de precisión para cada umbral.
    recalls: valores de recall para cada umbral.
    thresholds: diferentes umbrales que utiliza el modelo.
    ¿Por qué utilizamos y_cv_proba[:, 1]?
    Porque queremos utilizar las probabilidades de que cada paciente tenga una enfermedad cardíaca (clase positiva).
"""

""" 
    Si reducimos el threshold, el modelo clasifica más pacientes como enfermos. 
    Así detectamos más pacientes que realmente están enfermos (aumenta el recall), 
    pero también podemos cometer más falsos positivos (disminuye la precision).
"""

# Plot the results
plt.figure(figsize=(10, 6)) 
plot_precision_recall_vs_threshold(precisions, recalls, thresholds) 
plt.title("Precision and Recall vs Threshold") 

plt.grid() 

plt.show()
```

As we can see, the lower the probability threshold, the higher the recall and the lower the precision. There is always a cost: this is what is called the precision/recall tradeoff. A classic way of seeing that is to plot the so-called precision-recall curve, with recall as abcissa and precision in ordinate. Plot it :

```python
def plot_precision_recall_curve(precisions, recalls):
    plt.step(recalls, precisions, linewidth=2)
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    
#TOFILL
plt.figure(figsize=(8, 6))

plot_precision_recall_curve(precisions, recalls)

plt.title('Precision-Recall Curve')
plt.grid() 
plt.show()

""" 
    Pourquoi cette courbe est-elle importante ?

    Elle permet de visualiser le compromis entre les deux métriques :
        Si tu augmentes le Recall, tu détectes davantage de patients malades, mais tu risques d'avoir davantage de faux positifs.
        Si tu augmentes la Precision, tes prédictions positives sont plus fiables, mais tu risques de manquer certains patients malades.
"""
```

Here again, we can see the precision/recall trade-off. The better the classifier is, the closer the curve will be to the top-right corner. One way to summarize all these trade-offs in a single metric is to compute the area under the precision-recall curve, also called the *average precision* :

```python
from sklearn.metrics import average_precision_score

#TOFILL

## Calculate Average Precision

average_precision = average_precision_score(y_train, y_cv_proba[:, 1])

print("Average Precision:", average_precision) 
print(f"Average Precision: {average_precision * 100:.2f}%")

""" 
    ¿Cómo interpretar esta métrica?
        AP = 1: modelo perfecto.
        AP cercano a 1: muy buen rendimiento.
        AP cercano a 0: rendimiento muy bajo.
    Importante: cuanto mayor sea el Average Precision, mejor será el equilibrio entre Precision y Recall
    a través de los diferentes umbrales.
    No obstante, recuerda que esta métrica se calcula sobre tus predicciones de validación cruzada (y_cv_proba),
    no sobre el conjunto de test.
"""

```

The average precision is generally the preferred metric if you can compute your test samples probabiliies, and if you care more about the positive class and/or have an imbalanced dataset.

So, let's not forget we are trying to diagnose people's heart disease. Let's say we are ready to accept to incorrectly label as healthy at most 5% of the patients that have a heart disease (false negatives), i.e. we want a recall of 0.95.
Given the threshold curve above, choose a threshold that approximately yields a recall of 0.95, and recompute the confusion matrix as well as precision:

```python
"""  
    Recall = 0.95 → detectamos el 95 % de los pacientes enfermos.
    False Negative Rate = 0.05 → el 5 % de los pacientes enfermos son clasificados incorrectamente como sanos.
    Para conseguirlo, debemos reducir el threshold respecto al valor habitual de 0.5.
"""
#TOFILL
import numpy as np

"""
    recalls[:-1]: selecciona los valores de Recall correspondientes a los thresholds.
    np.abs(recalls[:-1] - 0.95): calcula la distancia entre cada Recall y 0.95.
    np.argmin(): encuentra la posición del valor más cercano.
    thresholds[idx]: recupera el threshold correspondiente. 
    
    utilizamos y_cv_proba[:, 1], porque queremos las probabilidades de tener una enfermedad cardíaca (clase positiva).
"""

#TOKEEP: threshold = , y_cv_pred_95recall = y_cv_proba > threshold
# Find the threshold that gives a recall closest to 0.95
idx = np.argmin(np.abs(recalls[:-1] - 0.95))

threshold = thresholds[idx]

# Generate new predictions
#y_cv_pred_95recall = y_cv_proba >= threshold
y_cv_pred_95recall = (y_cv_proba[:, 1] >= threshold).astype(int)

print(y_cv_pred_95recall)


print("Selected threshold:", threshold)
print("Corresponding recall:", recalls[idx])


```

## The Receiver Operating Characteristic (ROC) curve

(Yet) another way to look at your classfier performance, is to consider the false positive rate (FPR) instead of precision:


$$FPR= \frac{FP}{FP + TN}$$

This is the proportion of negative test samples that are incorrectly classified as positive. Plotting the FPR against the recall gives another classic evaluation curve, the ROC curve. 

```python
#The ROC curve
from sklearn.metrics import roc_curve

""" 
la curva ROC (Receiver Operating Characteristic).

    Esta curva permite evaluar el rendimiento de nuestro clasificador KNN observando la relación entre dos métricas:

        ** FPR (False Positive Rate): proporción de pacientes sanos que el modelo clasifica incorrectamente como enfermos.
        ** TPR (True Positive Rate): proporción de pacientes enfermos que el modelo identifica correctamente. Es exactamente 
        el Recall.
"""


def plot_roc_curve(fpr, tpr, label=None):
    plt.plot(fpr, tpr, linewidth=2, label=label)
    plt.plot([0, 1], [0, 1], 'k--')
    plt.axis([0, 1, 0, 1])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate (=Recall)')
    
""" 
    Variable	    Significado
    fpr	            Tasa de falsos positivos.
    tpr	            Tasa de verdaderos positivos (Recall).
    thresholds_roc	Diferentes umbrales de clasificación.

    Utilizamos y_cv_proba[:, 1] porque queremos las probabilidades de que un paciente tenga una enfermedad cardíaca.
"""

#TOFILL
# Calculate FPR, TPR and thresholds
fpr, tpr, thresholds_roc = roc_curve( y_train, y_cv_proba[:, 1] )

# Plot ROC curve
plt.figure(figsize=(8, 6)) 
plot_roc_curve(fpr, tpr)

plt.title("ROC Curve") 
plt.grid() 
plt.show()


"""  
¿Qué buscamos?
    Queremos que nuestra curva se aproxime lo máximo posible a la esquina superior izquierda:

    FPR = 0 → ningún falso positivo.
    TPR = 1 → detectamos todos los pacientes enfermos.
importante: la curva ROC nos permite estudiar la capacidad del modelo para 
distinguir entre pacientes sanos y enfermos, 
teniendo en cuenta los diferentes umbrales de clasificación.
"""
```

Here again there is a trade-off between TPR and FPR. Again, we can compute the area under this curve (AUC) :

```python
# AUC (Area Under the Curve) de la curva ROC, es decir, el área bajo la curva
from sklearn.metrics import roc_auc_score

#TOFILL
## Calculate ROC AUC

roc_auc = roc_auc_score(y_train, y_cv_proba[:, 1])

""" 
¿Cómo interpretar el AUC?
    AUC	        Interpretación
    1.0	        Clasificador perfecto.
    0.9 - 1.0	Excelente.
    0.8 - 0.9	Bueno.
    0.7 - 0.8	Aceptable.
    0.5 - 0.7	Capacidad de discriminación limitada.
    0.5	        Equivalente a una clasificación aleatoria.

    Recuerda: cuanto más se aproxime el AUC a 1, mejor será la capacidad de discriminación del clasificador.
"""

print("ROC AUC:", roc_auc) 
print(f"ROC AUC: {roc_auc * 100:.2f}%")

```

The ROC-AUC is useful when you can compute your test samples probabilities andcare about how well ranked are your predictions, but it should be used when you have a balanced dataset and care equally about both targets, so we won't use it in the following because we care more about the positive class.

## Hyper-parameters search : the best number of neighbors

Now we know that in our case (we care more about the people who are sick, i.e. the positive class), the best way to compare classifiers with a single metric is *average precision* (AP), so we can start choosing the best hyper-parameter k. You have already studied the bias/variance tradeoff with linear regression. Here this is the same: with k=1 the bias will be 0, but the variance very high, and conversely with too high a k. In both cases, we will have a sub-optimal AP.

You can read more about the bias/variance trade-off with KNNs here: https://medium.com/30-days-of-machine-learning/day-3-k-nearest-neighbors-and-bias-variance-tradeoff-75f84d515bdb


Let's search for the value of k that gives the best AP score, by doing a grid-search as you already did with linear regression, but first lets play a bit with the value of k. As you have already learnt, to choose the best hyper-parameters, we leave the test set aside, and do a *cross-validation* on the train set, from which we can compute *validation metrics* for each hyper-parameter combination to select them. 

Plot the PR curve and compute the mean validation APs with k=15, 30 and 50 by doing a 5-fold cross-validation over the training set :

```python
from sklearn.model_selection import cross_val_predict
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import precision_recall_curve, average_precision_score

#TOFILL

## Values of k to compare
neighbors = [15, 30, 50]



# Create KNN classifier
# creando un clasificador KNN (K-Nearest Neighbors) y especificando cuántos vecinos debe tener en cuenta para realizar una predicción.
# n_neighbors :Es un hiperparámetro de KNN que indica el número de vecinos más cercanos que el modelo utilizará para clasificar un nuevo paciente.
# k es una variable que va tomando los valores de la lista neighbors.
for k in neighbors:
    knn_clf = KNeighborsClassifier(n_neighbors=k) 

""" 
¿Por qué probar diferentes valores?
    Porque queremos encontrar el número de vecinos que proporciona el mejor rendimiento.
        Pocos vecinos → modelo más sensible a los datos individuales (mayor varianza).
        Muchos vecinos → modelo más generalizado, pero puede perder capacidad para detectar 
        patrones específicos (mayor sesgo).
"""

# Cross-validation predictions (probabilities)
y_cv_proba_k = cross_val_predict(
    knn_clf,
    X_train_scaled,
    y_train,
    cv=5,
    method='predict_proba'
)

# Calculate Precision and Recall
precisions_k, recalls_k, thresholds_k = precision_recall_curve(
    y_train,
    y_cv_proba_k[:, 1]
)

# Calculate Average Precision
ap = average_precision_score(y_train, y_cv_proba_k[:, 1])


plt.figure(figsize=(10, 7))

# Plot Precision-Recall curve
plt.plot(recalls_k, precisions_k, label=f'k={k}, AP={ap:.3f}')

print(f"k = {k} --> Average Precision = {ap:.4f}")

## Configure graph
plt.xlabel("Recall") 
plt.ylabel("Precision") 
plt.title("Precision-Recall Curves for Different KNN Neighbors") 
plt.legend(loc="lower left") 
plt.grid() 
plt.show()


```
